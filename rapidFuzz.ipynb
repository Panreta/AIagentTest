```python
from rapidfuzz import fuzz
print(fuzz.ratio("this is a test", "this is a test!"))

print(fuzz.partial_ratio("this is a test", "this is a test!")) # partial, ignore

```

Token Sort Ratio: The words are right that's ok

```python
fuzz.token_sort_ratio("fuzzy wuzzy was a bear", "wuzzy fuzzy was a bear")
```

Token Set Ratio: ignore frequency

```python
print(fuzz.token_sort_ratio("fuzzy wuzzy was a bear", "wuzzy fuzzy was a bear"))
print(fuzz.token_set_ratio("fuzzy was a bear but not a dog", "fuzzy was a bear but not a cat"))
```

Weighted Ratio

```python
print(fuzz.WRatio("this is a test", "this is a new test!!!"))
print()
```

```python

import json
import re

raw_text = '{"example": "format"} now here is my real answer {"score": 0.9}'
match = re.search(r'\{.*\}', raw_text, re.DOTALL)
if not match:
    raise RuntimeError(f"No JSON object found in model output: {raw_text!r}")

json_str = match.group(0)
print(f"Extracted JSON string: {json_str}")
try:
    parsed = json.loads(json_str)
except json.JSONDecodeError as e:
    raise RuntimeError(f"Failed to parse JSON: {json_str!r}") from e
```
